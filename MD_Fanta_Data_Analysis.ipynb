## 3. Standardize the data before analysis

```python
# Remove extra spaces and keep text values in a consistent form.
text_cols = df.select_dtypes(include='object').columns
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

df['countries_tags'] = df['countries_tags'].str.lower()
df = df.replace(['nan', 'NaN', '', 'undefined', 'UNKNOWN'], 'Unknown')

print('Data is cleaned and standardized.')
print(df['countries_tags'].value_counts())
```

### Note on missing values

`Unknown` is kept in the dataframe so that records are not deleted. When a chart or comparison is made, `Unknown` is left out because it does not tell us anything about the product itself.

## 4. Number of products by country

```python
country_counts = df['countries_tags'].value_counts().reindex(
    ['united-states','italy','united-kingdom','australia','romania']
)
print(country_counts)

plt.figure(figsize=(9,5))
sb.barplot(x=country_counts.index, y=country_counts.values)
plt.title('Number of Products by Country')
plt.xlabel('Country')
plt.ylabel('Number of Products')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Key Insights — Number of Products by Country

- **United States** has the highest number of products, with approximately **435 products**.
- **Italy** ranks second with around **315 products**.
- **United Kingdom** has approximately **145 products**, significantly lower than the top two countries.
- **Australia** has around **95 products**.
- **Romania** has the lowest product count, with approximately **70 products**.

## Overall Observation

> **Product distribution is highly concentrated in the United States and Italy**, while the remaining countries have considerably fewer products.
