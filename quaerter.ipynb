```python
pp_for_plot = pp[pp != 'unknown']
plt.figure(figsize=(14, 8))
ax = sb.countplot(x=pp_for_plot)
ax.set_title('Purchase Place Counts')
ax.set_xlabel('Purchase Place')
ax.set_ylabel('Count')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
top15_pp = pp_for_plot.value_counts().nlargest(15).index
top15_pp_data = pp_for_plot[pp_for_plot.isin(top15_pp)]

plt.figure(figsize=(16, 8))
ax = sb.countplot(x=top15_pp_data, order=top15_pp)
ax.set_title('Top 15 Purchase Places')
ax.set_xlabel('Purchase Place')
ax.set_ylabel('Count')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.tight_layout()
plt.show()
```
