# france-Focused Data Extraction, Cleaning & Exploratory Analysis

```python
france_data = df[
    df['countries_tags'].str.contains('france', case=False, na=False)
][[
    'product_name',
    'nutriscore_grade',
    'nutriments.sugars_100g',
    'nutriments.energy-kcal_100g',
    'packaging_tags',
    'purchase_places',
    'categories_tags',
    'nova_group',
    'countries_tags'
]]

france_data
```

```python
france_data.info()
```

```python
packaging = france_data.copy()

def combine_packaging(x):
    x = str(x).lower().strip()

    if x == 'unknown' or x == 'nan' or x == '':
        return 'unknown'

    elif 'plastic' in x or 'plastique' in x or 'kunststoff' in x:
        return 'plastic'

    elif 'bottle' in x or 'bouteille' in x or 'flasche' in x:
        return 'bottle'

    elif 'can' in x or 'aluminium' in x or 'metal' in x:
        return 'can'

    elif 'cardboard' in x or 'paperboard' in x or 'papp' in x:
        return 'cardboard'

    elif 'paper' in x:
        return 'paper'

    elif 'glass' in x:
        return 'glass'

    elif 'bag' in x:
        return 'bag'

    else:
        return 'other'

packaging['packaging_type'] = packaging['packaging_tags'].apply(combine_packaging)

packaging_clean = packaging[
    packaging['packaging_type'] != 'unknown'
].copy()

packaging_counts = packaging_clean['packaging_type'].value_counts()

ax = packaging_counts.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Packaging Types in France', fontsize=16, fontweight='bold')
plt.xlabel('Packaging Type', fontsize=12)
plt.ylabel('Number of Products', fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.bar_label(ax.containers[0], fontsize=10, padding=3)

plt.tight_layout()
plt.show()
```

### Observation: Packaging Type Distribution in France

- **Plastic packaging** is the most common type, with **127 products**, showing a strong preference for plastic packaging.
- **Can packaging** is the second most common, with **21 products**.
- **Cardboard** and **other packaging** each account for **18 products**.
- **Bottle packaging** is used for **12 products**, while **glass** is used for **7 products**.
- **Bag packaging** is found in **4 products**, and **paper** is the least common type with only **1 product**.
- Overall, the packaging distribution is **highly concentrated in plastic**, while paper and bag packaging have very limited representation.
