```python
france_data = df[
    df['countries_tags'].str.contains('france', case=False, na=False)
][[
    'product_name',
    'nutriscore_grade',
    'nutriments.sugars_100g',
    'nutriments.energy-kcal_100g',
    'packaging_tags',
    'purchase_places',
    'categories_tags',
    'nova_group',
    'countries_tags'
]]

france_data
```

```python
france_data.info()
```

```python
df.packaging_tags.value_counts()
```

```python
packaging = france_data['packaging_tags'].str.lower().str.strip()

packaging = packaging.replace({
    # Bag
    r'.*bag.*': 'bag',

    # Bottle
    r'.*bottle.*': 'bottle',

    # Can
    r'.*can.*': 'can',

    # Plastic
    r'.*plastic.*': 'plastic',

    # Paper
    r'.*paper.*': 'paper',

    # Metal
    r'.*metal.*|.*aluminium.*': 'metal',

    # Glass
    r'.*glass.*': 'glass'
}, regex=True)

packaging.value_counts()
```

```python
packaging = france_data[france_data['packaging_tags'] != 'unknown']

packaging['packaging_tags'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Packaging Types")
plt.xlabel("Packaging Type")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

### Observation-Analysis of the Top 10 Packaging Types by Product Count



- The chart displays the **Top 10 Packaging Types** based on their occurrence in the dataset.
- **Plastic bag** is the most common packaging type, with approximately **28 products**, making it the dominant packaging material.
- **Plastic** and **Plastic bottle** are the second most frequent packaging types, each appearing around **16 times**.
- **Cardboard** ranks next with approximately **11 products**, indicating its importance as an alternative packaging material.
- Packaging types such as **Bottle-plastique**, **Can (drink-can)**, **Plastic tray**, **Glass jar**, **Box (cardboard)**, and **Bottle** occur less frequently, each appearing fewer than **10 times**.
- Overall, the analysis shows that **plastic-based packaging dominates the dataset**, while paper-, glass-, and metal-based packaging types are comparatively less common.

```python
nutri = france_data[(france_data['nutriscore_grade'] != 'unknown') & (france_data['nutriscore_grade'].notna())]

nutri['nutriscore_grade'].value_counts().sort_index().plot(kind='bar')
plt.title("Nutri-Score Distribution")
plt.xlabel("Nutri-Score")
plt.ylabel("Number of Products")
plt.show()
```

### Observation-Analysis of Nutri-Score Distribution



- The chart illustrates the distribution of products across different **Nutri-Score** grades.

- **Nutri-Score D** is the second most common grade, followed by **C**, while **B** and **A** contain relatively few products.
- The distribution is highly skewed toward lower nutritional grades, with a significant gap between **E** and all other categories.
- Very few products are marked as **Not Applicable**, suggesting that Nutri-Score information is available for the majority of products.
- Overall, the dataset indicates that a large proportion of products have **lower nutritional quality**, highlighting opportunities for healthier product choices and reformulation.
