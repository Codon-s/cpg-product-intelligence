```python

# 1. Clean and explode safely (keeping rows aligned)
df_clean = df[['countries_tags', 'nutriscore_grade']].copy()
df_clean['countries_tags'] = df_clean['countries_tags'].astype(str).str.split(',')
df_clean = df_clean.explode('countries_tags')
df_clean['countries_tags'] = df_clean['countries_tags'].str.strip()

# 2. Identify unique Top 10 Countries 
top_10_countries = df_clean['countries_tags'].value_counts().head(10).index.drop_duplicates()

# 3. Filter for Top 10 Countries AND drop 'unknown' scores
df_mixed = df_clean[
    (df_clean['countries_tags'].isin(top_10_countries)) & 
    (df_clean['nutriscore_grade'] != 'unknown')
]

# 4. Create the crosstab
combined_data = pd.crosstab(
    index=df_mixed['countries_tags'],
    columns=df_mixed['nutriscore_grade']
)

# 5. Fix Duplicate Reindexing Error: Safely order by top 10
valid_countries = [c for c in top_10_countries if c in combined_data.index]
combined_data = combined_data.loc[valid_countries]

# 6. Plot the graph
combined_data.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('Top 10 Countries Mixed with Nutri-Score Grade')
plt.xlabel('Countries')
plt.ylabel('Product Count')
plt.legend(title='Nutri-Score')
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Prepare data for Countries vs NOVA Group
df_nova = df[['countries_tags', 'nova_group']].copy()
df_nova['countries_tags'] = df_nova['countries_tags'].astype(str).str.split(',')
df_nova = df_nova.explode('countries_tags')
df_nova['countries_tags'] = df_nova['countries_tags'].str.strip()

# 2. Get Top 10 Countries
top_10_countries = df_nova['countries_tags'].value_counts().head(10).index.drop_duplicates()

# 3. Filter for Top 10 and remove missing NOVA scores
df_nova_mixed = df_nova[
    (df_nova['countries_tags'].isin(top_10_countries)) & 
    (df_nova['nova_group'].notna())
]

# 4. Create the mix (Crosstab)
nova_data = pd.crosstab(
    index=df_nova_mixed['countries_tags'],
    columns=df_nova_mixed['nova_group']
)

# 5. Safely sort by Top 10 order
valid_countries = [c for c in top_10_countries if c in nova_data.index]
nova_data = nova_data.loc[valid_countries]

# 6. Plot the graph
nova_data.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

plt.title('Top 10 Countries Mixed with NOVA Group (Processing Level)')
plt.xlabel('Countries')
plt.ylabel('Product Count')
plt.legend(title='NOVA Group (1=Unprocessed, 4=Ultra)')
plt.xticks(rotation=45)
plt.show()
```

```python
df_pkg = df[['packaging_tags', 'nutriscore_grade']].copy()
df_pkg['packaging_tags'] = df_pkg['packaging_tags'].astype(str).str.split(',')
df_pkg = df_pkg.explode('packaging_tags')
df_pkg['packaging_tags'] = df_pkg['packaging_tags'].str.strip()

# Exclude 'unknown', 'nan', and empty strings from packaging tags
df_pkg = df_pkg[~df_pkg['packaging_tags'].isin(['unknown', 'nan', ''])]

# 2. Get Top 10 Packaging Tags
top_10_packaging = df_pkg['packaging_tags'].value_counts().head(10).index.drop_duplicates()

# 3. Filter for Top 10 Packaging
df_pkg_mixed = df_pkg[df_pkg['packaging_tags'].isin(top_10_packaging)]

# 4. Create the mix (Crosstab)
pkg_nutri_data = pd.crosstab(
    index=df_pkg_mixed['packaging_tags'],
    columns=df_pkg_mixed['nutriscore_grade']
)

# 5. Explicitly drop the 'unknown' column if it exists in Nutri-Score
pkg_nutri_data = pkg_nutri_data.drop(columns=['unknown'], errors='ignore')

# 6. Safely sort by Top 10 order
valid_packaging = [p for p in top_10_packaging if p in pkg_nutri_data.index]
pkg_nutri_data = pkg_nutri_data.loc[valid_packaging]

# 7. Plot the graph
pkg_nutri_data.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('Top 10 Packaging Tags Mixed with Nutri-Score Grade')
plt.xlabel('Packaging Tags')
plt.ylabel('Product Count')
plt.legend(title='Nutri-Score')
plt.xticks(rotation=45)
plt.show()
```

```python
df_pkg = df[['packaging_tags', 'nova_group']].copy()
df_pkg['packaging_tags'] = df_pkg['packaging_tags'].astype(str).str.split(',')
df_pkg = df_pkg.explode('packaging_tags')
df_pkg['packaging_tags'] = df_pkg['packaging_tags'].str.strip()

# Exclude 'unknown', 'nan', and empty strings
df_pkg = df_pkg[~df_pkg['packaging_tags'].isin(['unknown', 'nan', ''])]

# 2. Get Top 10 Packaging Tags
top_10_packaging = df_pkg['packaging_tags'].value_counts().head(10).index.drop_duplicates()

# 3. Filter for Top 10 Packaging and valid NOVA group values
df_pkg_mixed = df_pkg[
    (df_pkg['packaging_tags'].isin(top_10_packaging)) & 
    (df_pkg['nova_group'].notna())
]

# 4. Create the mix (Crosstab)
pkg_nova_data = pd.crosstab(
    index=df_pkg_mixed['packaging_tags'],
    columns=df_pkg_mixed['nova_group']
)

# 5. Safely sort by Top 10 order
valid_packaging = [p for p in top_10_packaging if p in pkg_nova_data.index]
pkg_nova_data = pkg_nova_data.loc[valid_packaging]

# 6. Plot the graph
pkg_nova_data.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

plt.title('Top 10 Packaging Tags Mixed with NOVA Group')
plt.xlabel('Packaging Tags')
plt.ylabel('Product Count')
plt.legend(title='NOVA Group (1=Unprocessed, 4=Ultra)')
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Clean and explode safely using the DataFrame explode method to prevent index errors
df_cp = df[['countries_tags', 'packaging_tags']].copy()

df_cp['countries_tags'] = df_cp['countries_tags'].astype(str).str.split(',')
df_cp = df_cp.explode('countries_tags')
df_cp['countries_tags'] = df_cp['countries_tags'].str.strip()

df_cp['packaging_tags'] = df_cp['packaging_tags'].astype(str).str.split(',')
df_cp = df_cp.explode('packaging_tags')
df_cp['packaging_tags'] = df_cp['packaging_tags'].str.strip()

# Filter out unknown or empty values
df_cp = df_cp[
    (~df_cp['countries_tags'].isin(['unknown', 'nan', ''])) & 
    (~df_cp['packaging_tags'].isin(['unknown', 'nan', '']))
]

# 2. Get Top 5 Countries and Top 5 Packaging Tags
top_countries = df_cp['countries_tags'].value_counts().head(5).index.drop_duplicates()
top_packaging = df_cp['packaging_tags'].value_counts().head(5).index.drop_duplicates()

df_cp_mixed = df_cp[
    (df_cp['countries_tags'].isin(top_countries)) & 
    (df_cp['packaging_tags'].isin(top_packaging))
]

# 3. Create Crosstab
cp_data = pd.crosstab(
    index=df_cp_mixed['countries_tags'],
    columns=df_cp_mixed['packaging_tags']
)

# Safely sort by top countries order
valid_countries = [c for c in top_countries if c in cp_data.index]
cp_data = cp_data.loc[valid_countries]

# 4. Plot the graph
cp_data.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('Top 5 Countries Mixed with Top 5 Packaging Tags')
plt.xlabel('Countries')
plt.ylabel('Product Count')
plt.legend(title='Packaging Tag')
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Filter valid NOVA and Nutri-Score data
df_pct = df[
    (df['nova_group'].notna()) & 
    (df['nutriscore_grade'].isin(['unknown', 'nan', ''])) &
    (df['nutriscore_grade'].notna())
].copy()

# 2. Create the Crosstab
pct_data = pd.crosstab(
    index=df_pct['nova_group'],
    columns=df_pct['nutriscore_grade']
)

# 3. Convert to 100% Percentages for equal comparison
pct_data = pct_data.div(pct_data.sum(axis=1), axis=0) * 100

# 4. Plot 100% Stacked Bar Graph
pct_data.plot(kind='bar', stacked=True, figsize=(10, 6))

plt.title('100% Proportion of Nutri-Scores within each NOVA Group')
plt.xlabel('NOVA Group (1=Unprocessed, 4=Ultra)')
plt.ylabel('Percentage (%)')
plt.legend(title='Nutri-Score', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
