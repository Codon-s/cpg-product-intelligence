```python
# ============================================================
# PART 0: LOAD AND PREPARE THE CLEANED FIVE-COUNTRY DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter

# Load the ONE cleaned CSV
df = pd.read_csv('datasets/fanta_5_countries_cleaned.csv')

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ============================================================
# CHECK COUNTRY DISTRIBUTION
# ============================================================

print("Country Distribution:")
display(df["countries_tags"].value_counts())

print("\nCountry Percentage:")
country_percentage = (
    df["countries_tags"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

display(country_percentage)
```

```python
# ============================================================
# NUMERIC COLUMNS
# ============================================================

numeric_columns = [
    "nova_group",
    "nutriments.sugars_100g",
    "nutriments.energy-kcal_100g"
]

for col in numeric_columns:
    df[col + "_num"] = pd.to_numeric(df[col], errors="coerce")

print("Numeric analytical columns created:")
display(df[
    [
        "nova_group_num",
        "nutriments.sugars_100g_num",
        "nutriments.energy-kcal_100g_num"
    ]
].head())
```
