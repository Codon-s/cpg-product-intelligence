```python
ingredients = (
    us_df['ingredients_text']
    .dropna()
    .astype(str)
    .str.split(',')
    .explode()
    .str.strip()
    .str.lower()
    .str.rstrip('.')
    .str.strip()
)

unknown_count = (ingredients == 'unknown').sum()
total_ingredients = len(ingredients)

unknown_proportion = unknown_count / total_ingredients
unknown_percentage = unknown_proportion * 100

print("Unknown data count:", unknown_count)
print("Total ingredient entries:", total_ingredients)
print("Proportion of unknown data:", round(unknown_proportion, 4))
print("Percentage of unknown data:", round(unknown_percentage, 2), "%")


top_ingredients = (
    ingredients[
        (ingredients != 'unknown') &
        (~ingredients.str.match(r'^\d+$')) &
        (~ingredients.str.contains('beverage drive', case=False, na=False))
    ]
    .value_counts()
    .head(5)
)

plt.figure(figsize=(12,8))

sb.barplot(y=top_ingredients.index, x=top_ingredients.values)

plt.xlabel("Count")
plt.ylabel("Ingredient")
plt.title("Top Ingredients in United States Products")

plt.tight_layout()
plt.show()
```

Observation: Carbonated water is the most common ingredient, appearing approximately 28 times, followed by natural flavors with around 25 occurrences. Citric acid, high fructose corn syrup, sodium benzoate, and caramel color also appear frequently, indicating the regular use of sweeteners, preservatives, acids, and coloring agents. Other ingredients such as quinine, aspartame, Red 40, Blue 1, sodium bicarbonate, and sodium chloride occur less frequently, generally around 2–4 times.

Interpretation: The ingredient distribution is mainly dominated by carbonated water and flavoring ingredients, while sweeteners, preservatives, acids, and color additives are also commonly present across the products.

Section-ready takeaway: Carbonated water and natural flavors are the most frequently occurring ingredients, while sweeteners, preservatives, acids, and color additives also show a strong presence in the products.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

countries = {
    'United States': us_df,
    'Australia': aus_data,
    'Germany': germany_data,
    'Belgium': belgium_data,
    'Switzerland': switzerland_data
}

all_calories = []

for country, data in countries.items():

    calories = (
        pd.to_numeric(
            data['nutriments.energy-kcal_100g'],
            errors='coerce'
        )
        .dropna()
    )

    avg_calories = calories.mean()

    all_calories.append({
        'Country': country,
        'Average Calories': avg_calories
    })

calories_graph = pd.DataFrame(all_calories)

plt.figure(figsize=(12, 7))

sb.barplot(
    data=calories_graph,
    x='Country',
    y='Average Calories'
)

plt.xlabel("Country")
plt.ylabel("Average Energy (kcal/100g)")
plt.title("Average Energy Content Across Countries")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

Observation :-

The average energy content is highest in the United States at around 77 kcal/100g, while Australia has a much lower average of around 30 kcal/100g. Germany, Belgium, and Switzerland have no visible values in this graph, suggesting that the available energy data for these countries was insufficient.
